## 🔢 ¿Por qué los índices en Python comienzan desde 0?

Esto ocurre por razones **técnicas** y **conceptuales**:

- **Técnica:** en memoria, cada elemento de una lista ocupa una posición consecutiva.  
  El índice indica *cuántos pasos hay desde el inicio*, no el número del elemento.

- **Conceptual:** el índice `0` significa “no te muevas del comienzo”.  
  Por eso, el primer elemento tiene desplazamiento `0`, el segundo `1`, el tercero `2`, y así sucesivamente.

- **Ventaja práctica:** esta convención simplifica muchas operaciones, como el corte de listas (`slicing`), donde el rango se interpreta como `[inicio:fin)` —es decir, incluye el inicio pero **excluye el final**.

### 5.3 🎯 `.loc` vs `.iloc` en pandas — explicación breve

**Idea clave**  
- **`.loc[fila, columna]`** selecciona por **etiquetas** (nombres del índice y de las columnas).  
- **`.iloc[fila, columna]`** selecciona por **posición** (enteros 0, 1, 2, …).  

**Diferencias importantes**  
- Con **`.loc`** los **rangos** por etiqueta son **inclusivos** en el final.  
- Con **`.iloc`** los **rangos** por posición son **exclusivos** en el final (como `range` de Python).  
- Para **asignar/editar** usa siempre `.loc[...] = ...` (evita `SettingWithCopyWarning`).

### 5.6 🔗 Uniones y combinaciones en pandas: `merge` y `concat`

### 🤝 `merge`

Cuando haces `merge(left, right, on=..., how=...)`, pandas **combina columnas** de dos tablas
según una **clave** (o varias). Lo importante es **qué filas conserva** cada tipo de `how`.

### 🧭 Reglas de conservación de filas

| `how`   | ¿Qué filas conserva? | ¿Cuándo usarlo? |
|---------|-----------------------|-----------------|
| `inner` | **Solo** las claves presentes en **ambas** tablas. | Cuando quieres la **intersección**; asegurar que hay correspondencia en los dos lados. |
| `left`  | **Todas** las filas de `left`; completa con `NaN` lo que falte de `right`. | Cuando `left` es tu **base principal** y **no quieres perder** sus filas. |
| `right` | **Todas** las filas de `right`; completa con `NaN` lo que falte de `left`. | Al revés del `left`; `right` es la base prioritaria. |
| `outer` | **Todas** las claves de ambos lados; donde no hay match, `NaN`. | Para una **unión total** y luego analizar faltantes/desalineaciones. |
