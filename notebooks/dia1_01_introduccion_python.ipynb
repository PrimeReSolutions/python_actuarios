## 2. Condiciones, bucles y funciones

### 2.1 Condiciones

`if`, `elif` y `else` ejecutan un bloque u otro según una condición. En Python los bloques se delimitan con la **sangría** (cuatro espacios), no con palabras de cierre: todo lo que está sangrado debajo de un `if` pertenece a él.

```python
siniestralidad = 0.78

if siniestralidad > 1:
    print("Pérdida técnica")
elif siniestralidad > 0.7:
    print("Vigilar")
else:
    print("Dentro de lo esperado")
```

### 2.2 Bucles

`for` recorre los elementos de una colección. `enumerate` devuelve además la posición de cada elemento; con `start=1` la numeración empieza en 1, que es lo habitual para los periodos de un flujo.

```python
flujos = [1000, 1000, 1000]

for t, flujo in enumerate(flujos, start=1):
    print(f"Año {t}: {flujo}")
```

### 2.3 Funciones

Una función encapsula un cálculo para reutilizarlo. Se define con `def`, recibe argumentos y devuelve un resultado con `return`. El texto entre triples comillas documenta qué hace.

El factor de descuento a `t` años con tipo de interés `i` es $v^t = (1 + i)^{-t}$.

```python
def factor_descuento(tasa, t):
    """Factor de descuento v^t = (1 + tasa)^(-t)."""
    return (1 + tasa) ** (-t)

factor_descuento(0.03, 5)
```

### 4.8 Combinar tablas con `merge`

`merge` cruza dos tablas por una o varias columnas clave, como `BUSCARV` en Excel pero devolviendo todas las columnas de la otra tabla. El argumento `how` decide qué filas se conservan:

| `how` | Filas que conserva | Uso habitual |
|---|---|---|
| `inner` | Solo las claves presentes en las dos tablas | Intersección |
| `left` | Todas las de la tabla izquierda; `NaN` donde no hay correspondencia | La tabla izquierda es la base principal |
| `right` | Todas las de la tabla derecha | Caso simétrico del anterior |
| `outer` | Todas las claves de ambas tablas | Detectar descuadres entre dos fuentes |

Preparamos dos tablas auxiliares. Observa que `"Moto"` no existe en `df` y que `origin_factor` solo cubre los años 1988 a 1992.
