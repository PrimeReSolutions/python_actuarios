## 5. Leer un mensaje de error

Cuando una celda falla, Python muestra un informe de error. Se lee **de abajo arriba**: la última línea indica el tipo de error y su causa; las líneas superiores muestran en qué punto del código ocurrió. Un error no estropea nada: basta con corregir la celda y volver a ejecutarla.

| Tipo de error | Qué significa | Causa habitual en el curso |
|---|---|---|
| `SyntaxError` | Python no entiende la instrucción | Un `***` sin completar; un paréntesis o unas comillas sin cerrar |
| `NameError` | Un nombre no está definido | No se ejecutó una celda anterior, o el entorno se reinició |
| `KeyError` | La columna o la clave no existe | Error al escribir el nombre de una columna; Python distingue mayúsculas y minúsculas |
| `TypeError` | Operación con un tipo de dato incompatible | Sumar un texto y un número |
| `FileNotFoundError` | No encuentra el archivo | No se ejecutó la celda de preparación del entorno |

La celda siguiente produce un error a propósito. Ejecútala y localiza en la última línea el tipo de error.

````python
prima = "1200"     # el valor llegó como texto, por ejemplo desde un CSV
prima + 300
````

### Ejercicio 0.2 Corregir el error

La prima está guardada como texto (`"1200"`) y Python no suma texto con números. Sustituye `***` por la función que convierte un texto en número entero. El resultado debe ser `1500`.

````python
***(prima) + 300
````

<details>
<summary>Solución</summary>

`int(prima) + 300`. La función `int()` convierte el texto `"1200"` en el número `1200`. Si el valor tuviera decimales se usaría `float()`.

</details>
